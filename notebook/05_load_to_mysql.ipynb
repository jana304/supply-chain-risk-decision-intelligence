```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus
from getpass import getpass

# Load cleaned dataset
df_sql = pd.read_csv(
    "../data/DataCoSupplyChain_Cleaned.csv"
)

# Convert date columns
df_sql["order date (DateOrders)"] = pd.to_datetime(
    df_sql["order date (DateOrders)"],
    errors="coerce"
)

df_sql["shipping date (DateOrders)"] = pd.to_datetime(
    df_sql["shipping date (DateOrders)"],
    errors="coerce"
)

print("Rows:", len(df_sql))
print("Columns:", len(df_sql))
```

```python
print("Rows:", df_sql.shape[0])
print("Columns:", df_sql.shape[1])
```

```python
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus
import getpass

password = getpass.getpass("Enter your MySQL root password: ")

engine = create_engine(
    "mysql+mysqlconnector://root:" 
    + quote_plus(password)
    + "@localhost:3306/supply_chain_risk"
)

with engine.connect() as conn:
    database = conn.execute(text("SELECT DATABASE()")).scalar()

print("Connected to:", database)
```

```python
df_sql.to_sql(
    "supply_chain_data",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

print("Dataset loaded successfully!")
```
