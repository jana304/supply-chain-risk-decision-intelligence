# Supply Chain Risk & Operations Decision Intelligence
## 01 - Data Cleaning & Data Understanding

```python
# ============================================
# 1. IMPORT LIBRARIES
# ============================================

import pandas as pd
import numpy as np


# ============================================
# 2. LOAD DATASET
# ============================================

df = pd.read_csv(
    "../data/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)


# ============================================
# 3. BASIC DATASET INFORMATION
# ============================================

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)


# ============================================
# 4. FIRST 5 ROWS
# ============================================

df.head()
```

```python
# ============================================
# 5. CHECK COLUMN NAMES
# ============================================

print("Column Names:")
print(df.columns.tolist())


# ============================================
# 6. CHECK DATA TYPES
# ============================================

print("\nData Types:")
print(df.dtypes)


# ============================================
# 7. CHECK MISSING VALUES
# ============================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================
# 8. CHECK DUPLICATE ROWS
# ============================================

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# ============================================
# 9. DATASET SUMMARY
# ============================================

summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum().values,
    "Missing %": (df.isnull().mean() * 100).round(2).values,
    "Unique Values": df.nunique().values
})

summary
```

```python
# ============================================
# 10. CHECK TARGET VARIABLE
# ============================================

print("Late Delivery Risk Counts:")
print(df["Late_delivery_risk"].value_counts())

print("\nLate Delivery Risk Percentage:")
print(
    df["Late_delivery_risk"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================
# 11. REMOVE UNNECESSARY & SENSITIVE COLUMNS
# ============================================

columns_to_remove = [
    # Customer personal information
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street",
    
    # Technical / unnecessary information
    "Product Description",
    "Product Image",
    
    # Mostly missing and not useful for our analysis
    "Order Zipcode"
]

df_clean = df.drop(columns=columns_to_remove)

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRemoved columns:")
print(columns_to_remove)
```

```python
# ============================================
# 12. CONVERT DATE COLUMNS
# ============================================

df_clean["order date (DateOrders)"] = pd.to_datetime(
    df_clean["order date (DateOrders)"],
    errors="coerce"
)

df_clean["shipping date (DateOrders)"] = pd.to_datetime(
    df_clean["shipping date (DateOrders)"],
    errors="coerce"
)

print("Order date type:", df_clean["order date (DateOrders)"].dtype)
print("Shipping date type:", df_clean["shipping date (DateOrders)"].dtype)
```

```python
# ============================================
# 13. VALIDATE DATE COLUMNS
# ============================================

print("Missing order dates:")
print(df_clean["order date (DateOrders)"].isnull().sum())

print("\nMissing shipping dates:")
print(df_clean["shipping date (DateOrders)"].isnull().sum())

print("\nOrder date range:")
print(
    df_clean["order date (DateOrders)"].min(),
    "to",
    df_clean["order date (DateOrders)"].max()
)

print("\nShipping date range:")
print(
    df_clean["shipping date (DateOrders)"].min(),
    "to",
    df_clean["shipping date (DateOrders)"].max()
)
```

```python
# ============================================
# 14. CHECK IMPORTANT NUMERIC COLUMNS
# ============================================

numeric_columns = [
    "Sales",
    "Order Profit Per Order",
    "Product Price",
    "Order Item Product Price",
    "Order Item Quantity",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

df_clean[numeric_columns].describe().T
```

```python
# ============================================
# 15. CHECK IMPORTANT CATEGORICAL VALUES
# ============================================

categorical_columns = [
    "Shipping Mode",
    "Delivery Status",
    "Order Status",
    "Market",
    "Customer Segment",
    "Department Name",
    "Order Region",
    "Type"
]

for column in categorical_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    print(df_clean[column].value_counts())
```

```python
# ============================================
# 16. SAVE CLEANED DATASET
# ============================================

df_clean.to_csv(
    "../data/DataCoSupplyChain_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Shape:", df_clean.shape)
```
